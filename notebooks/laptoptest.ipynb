```python
import numpy as np
import time
```

```python
start = time.perf_counter()
matrix_floats = np.random.rand(300, 26000)
end = time.perf_counter()
print(f"{end - start:.6f} seconds")
```

```python
print((matrix_floats @ matrix_floats.conj().T).shape)
start2 = time.perf_counter()
W_inv = np.linalg.lstsq(matrix_floats @ matrix_floats.conj().T, np.eye(300))
end2 = time.perf_counter()
print(f"{end2 - start2:.6f} seconds")
```
